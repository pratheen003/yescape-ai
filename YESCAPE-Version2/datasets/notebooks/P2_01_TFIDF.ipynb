```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer

import joblib
import os
```

```python
dataset_path = r"D:\YESCAPE-AI\YESCAPE-Version2\datasets\processed\processed_dataset_v2.csv"

df = pd.read_csv(dataset_path)

print("Dataset Loaded Successfully!")

print(df.shape)
```

```python
df.info()

df.head()
```

```python
X = df["clean_text"]

y = df["fraudulent"]

print("Feature Samples")

print(X.head())

print()

print("Labels")

print(y.value_counts())
```

```python
X_train, X_test, y_train, y_test = train_test_split(

    X,

    y,

    test_size=0.20,

    random_state=42,

    stratify=y

)

print("Training Samples :", len(X_train))

print("Testing Samples :", len(X_test))
```

```python
tfidf = TfidfVectorizer(

    stop_words="english",

    max_features=5000,

    ngram_range=(1,1)

)

X_train_tfidf = tfidf.fit_transform(X_train)

X_test_tfidf = tfidf.transform(X_test)

print("TF-IDF Completed!")
```

```python
print("Training Matrix Shape")

print(X_train_tfidf.shape)

print()

print("Testing Matrix Shape")

print(X_test_tfidf.shape)
```

```python
feature_names = tfidf.get_feature_names_out()

print("Total Vocabulary Size")

print(len(feature_names))

print()

print(feature_names[:50])
```

```python
vectorizer_path = r"D:\YESCAPE-AI\YESCAPE-Version2\models\vectorizers\tfidf_vectorizer.pkl"

joblib.dump(

    tfidf,

    vectorizer_path

)

print("Vectorizer Saved Successfully!")
```

```python
split_path = r"D:\YESCAPE-AI\YESCAPE-Version2\models\vectorizers\train_test_split.npz"

np.savez(

    split_path,

    X_train_index=X_train.index,

    X_test_index=X_test.index,

    y_train=y_train,

    y_test=y_test

)

print("Train/Test Split Saved!")
```
