```python
df = pd.read_csv(
    r"D:\YESCAPE-AI\YESCAPE-Version2\datasets\processed\processed_dataset_v1.csv"
)

print(df.shape)
```

```python
text_columns = [
    "title",
    "company_profile",
    "description",
    "requirements",
    "benefits"
]

# Ensure every text feature is string
for col in text_columns:
    df[col] = df[col].fillna("").astype(str)

# Merge all text columns
df["combined_text"] = df[text_columns].agg(" ".join, axis=1)

print("Combined Text Created Successfully!")

df["combined_text"].head()
```

```python
print(df[text_columns].dtypes)
```

```python
df["combined_text"] = df["combined_text"].str.lower()
```

```python
def remove_html(text):

    return re.sub(r"<.*?>", "", text)

df["combined_text"] = df["combined_text"].apply(remove_html)
```

```python
def remove_urls(text):

    return re.sub(r"http\S+|www\S+", "", text)

df["combined_text"] = df["combined_text"].apply(remove_urls)
```

```python
def remove_numbers(text):

    return re.sub(r"\d+", "", text)

df["combined_text"] = df["combined_text"].apply(remove_numbers)
```

```python
def remove_special(text):

    return re.sub(r"[^a-zA-Z\s]", " ", text)

df["combined_text"] = df["combined_text"].apply(remove_special)
```

```python
def remove_spaces(text):

    return re.sub(r"\s+", " ", text).strip()

df["combined_text"] = df["combined_text"].apply(remove_spaces)
```

```python
import unicodedata

def normalize_unicode(text):
    return unicodedata.normalize("NFKD", text)

df["combined_text"] = df["combined_text"].apply(normalize_unicode)

print("Unicode Normalization Completed!")
```

```python
from nltk.tokenize import word_tokenize

df["tokens"] = df["combined_text"].apply(word_tokenize)
```

```python
stop_words = set(stopwords.words("english"))

df["tokens"] = df["tokens"].apply(
    lambda words: [
        word
        for word in words
        if word not in stop_words
    ]
)
```

```python
df["tokens"].iloc[0][:30]
```

```python
def remove_noise_words(tokens):

    clean_tokens = []

    for word in tokens:

        # Remove single-letter words
        if len(word) < 2:
            continue

        # Keep only alphabetic words
        if not word.isalpha():
            continue

        clean_tokens.append(word)

    return clean_tokens


df["tokens"] = df["tokens"].apply(remove_noise_words)

print("Noise words removed successfully!")
```

```python
lemmatizer = WordNetLemmatizer()

df["tokens"] = df["tokens"].apply(
    lambda words: [
        lemmatizer.lemmatize(word)
        for word in words
    ]
)
```

```python
df["clean_text"] = df["tokens"].apply(
    lambda words: " ".join(words)
)

df["clean_text"].head()
```
