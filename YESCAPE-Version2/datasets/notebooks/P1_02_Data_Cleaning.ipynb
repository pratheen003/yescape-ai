```python
import pandas as pd
import numpy as np

print("Libraries Imported Successfully!")
```

```python
emscad = pd.read_csv(
    r"D:\YESCAPE-AI\YESCAPE-Version2\datasets\base_datasets\emscad\DataSet.csv"
)

print(emscad.shape)
```

```python
df = emscad.copy()
```

```python
print("Before :", df.shape)

df = df.drop_duplicates()

print("After :", df.shape)
```

```python
df = df.drop(
    columns=[
        "department",
        "salary_range",
        "in_balanced_dataset"
    ]
)

print(df.columns)
```

```python
df.isnull().sum()
```

```python
text_columns = [

    "company_profile",

    "requirements",

    "benefits"

]

for col in text_columns:

    df[col] = df[col].fillna("")

metadata_columns = [

    "location",

    "employment_type",

    "required_experience",

    "required_education",

    "industry",

    "function"

]

for col in metadata_columns:

    df[col] = df[col].fillna("Unknown")
```

```python
df.isnull().sum()
```

```python
binary_columns = [
    "telecommuting",
    "has_company_logo",
    "has_questions",
    "fraudulent"
]

print("Checking Binary Columns...\n")

for col in binary_columns:
    print(f"{col}")
    print(df[col].value_counts())
    print("-" * 35)
```
